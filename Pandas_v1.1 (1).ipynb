```python
Oly_bio['born_datetime'] = pd.to_datetime(Oly_bio['born_date'])
```

```python
Oly_bio['born_year'] = Oly_bio['born_datetime'].dt.year
```

```python
Oly_bio[['name','born_year']]
```

```python
Oly_bio['height_Cat'] = Oly_bio['height_cm'].apply(lambda x: 'Short' if x<180 else 'Tall')
```

```python
Oly_bio
```

```python
Oly_bio['height_Cat'] = Oly_bio['height_cm'].apply(lambda x: 'Short' if x<180 else ('Average' if x<185 else 'Tall'))
```

```python
Oly_bio
```

```python
def categories_athlete(row):
    if row['height_cm'] <175 and row['weight_kg'] < 70:
        return 'Light_weight'
    elif row['height_cm'] <185 or row['weight_kg'] < 80:
        return 'Middle_weight'
    else:
        return 'Heavy weight'
```

```python
Oly_bio['Categories_athle'] = Oly_bio.apply(categories_athlete, axis =1)
```
